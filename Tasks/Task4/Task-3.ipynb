# Task 1.
- Calculate Factorial using recursion

```python
def factorial(num):
    if num == 0:
        return 0
    if num == 1:
        return 1
    return num * factorial(num -1 )

factorial(5)
```

```python
#### your class here
class fact:
    """
    Fact class handles the operation of calculating the factorial of a number using recursion
    Attributes:
    n (int): The number whose factorial will be calculated.
    """

    def __init__(self, n):
        """
        Initializes the Factorial object with the target number
        Args: n (int): The number whose factorial will be calculated
        """
        self.n = n

    def calculate(self):
        """
        calculates the factorial of the stored number using recursion
        Returns:
            int: The factorial of the number
        """
        return self.fact(self.n)

    def fact(self, n):
        """
        do the actual factorial computation
        Args:
        n(int): The current value in the recursion
        Returns:
            int: n! (n factorial).
        """
        if n == 0:
            return 1
        if n == 1:
            return 1
        return n * self.fact(n - 1)


#### test your class 
if __name__ == "__main__":
    # pass
    f = fact(5)
    print(f.calculate())
```

# Task 2.
- Write a function that takes a number and return True if this number is prime and False if not.

```python
def isPrime(num):
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

isPrime(20)
```

# Task 3.
- Write a function that takes two numbers and return a list of common dividors

```python
def dividors(num1, num2):
    mn = min(num1, num2)
    common_dividors = []
    for i in range(1, mn+1):
        if num1 % i == 0 and num2 % i ==0:
            common_dividors.append(i)
    return common_dividors

dividors(20, 10)
```

```python
#### your class here
class CommonDiv:
    """
    CommonDiv class handles the operation of finding all common divisors shared between two numbers
    Attributes:
        n1 (int): The first number
        n2 (int): The second number
    """
    def __init__(self, n1, n2):
        """
          initialize the CommonDiv object with the two numbers
        Args:
            n1 (int): The first number
            n2 (int): The second number
        """
        self.n1 = n1
        self.n2 = n2

    def find(self):
        """
        finds all common divisors of n1 and n2
        Returns:
        list: A list of integers that divide both n1 and n2
        """
        mn = min(self.n1, self.n2)
        common_divs = []
        for i in range(1, mn + 1):
            if self.n1 % i == 0 and self.n2 % i == 0:
                common_divs.append(i)
        return common_divs



#### test your class 
if __name__ == "__main__":
    # pass
    obj = CommonDiv(20, 10)
    print(obj.find())  
```

```python
#### your class here
class LongestCommonSubstring:
# finds the longest common substring between two strings
    def __init__(self, s1, s2):
        """
        initializes the two strings
        :param s1: First string
        :param s2: Second string
        """
        self.s1 = s1
        self.s2 = s2

    def find_longest_common(self):
        """
        finds the longest common substring
        :return: The longest common substring
        """
        longest_str = ""

        for i in range(len(self.s1)):
            for j in range(i, len(self.s1)):
                sub_str = self.s1[i:j+1]

                if sub_str in self.s2:
                    if len(sub_str) > len(longest_str):
                        longest_str = sub_str

        return longest_str


#### test your class
if __name__ == "__main__":
    #    pass
    s1 = input("Enter first string: ")
    s2 = input("Enter second string: ")

    obj = LongestCommonSubstring(s1, s2)

    print(obj.find_longest_common())



```

```python
#### your class here
class ReverseWords:
    """
    ReverseWords class reverses the order of words in a string
    Attributes:
        s (str): The input string
    """

    def __init__(self, s):
        """
        initializes the input string
        :param s: The input string
        """
        self.s = s

    def reverse(self):
        """
        reverses the order of the words in the string
        Returns:
        str: The string with the words reversed
        """
        words = self.s.split(" ")
        new_words = words[::-1]
        return " ".join(new_words)


#### test your class
if __name__ == "__main__":
    obj = ReverseWords("Hello World")
    print(obj.reverse())
```

- **Ex:** ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]  → {'Ali': 4, 'Welcome': 1, 'No': 3, 'Hi': 2}

```python
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
unique_words = set(words)
counts = {word: 0 for word in unique_words}
for i in unique_words:
    for j in words:
        if i == j:
            counts[i] += 1
counts
```

```python
#### your class here
class countWords:
    """
    countWords class counts the frequency of each unique word in a list
    Attributes:
    words (list):list of words
    """
    def __init__(self, words):
        """
        initializes the list of words
        :param
        words: The input list of words
        """
        self.words = words

    def count(self):
        """
        counts how many times each word appears
        Returns:
        dict:a dict containing each word and its frequency
        """
        unique_words = set(self.words)
        counts = {}

        for word in unique_words:
            counts[word] = 0

        for word in unique_words:
            for item in self.words:
                if word == item:
                    counts[word] += 1

        return counts


#### test your class
if __name__ == "__main__":
    words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
    obj = countWords(words)
    print(obj.count())
```

# Task 8.
- Write a Python program to find all the unique words and count the frequency of occurrence from a text file and save the counts in a new file
- This is the same as Task 7 but this time read words from a file
- Hint: use replace function and split function

**First Step: read the file**

```python

input_file = open("input_file.txt", "r")
text = input_file.read()
input_file.close()
```

**Second step: remove '\n' then convert the file text into a list of words**

```python
text = text.replace('\n', ' ')
words = text.split(' ')
```

**Third Step: get unique words, create a dictionary, then count unique words**

```python
unique_words = set(words)
word_counts = dict()

for word in unique_words:
    word_counts[word] = words.count(word)
```

**Save the word-count pair into a text file**

```python
output_file = open("output_file.txt", "w")
for word in word_counts:
    output_file.write("{}: {}\n".format(word, word_counts[word]))
output_file.close()
```

```python
#### your class here
class CountWordsFile:
    """
    counts the frequency of words in a text file
    Attributes:
        input (str): Input file name
        out (str): Output file name
    """

    def __init__(self, input, out):
        #stores the file names
        self.input_file = input
        self.output_file = out

    def count_words(self):
        """
        counts the words and saves the result
        Returns:
        dict: Word frequencies
        """
        file = open(self.input_file, "r")
        text = file.read()
        file.close()

        text = text.replace("\n", " ")
        words = text.split()

        unique_words = set(words)
        word_counts = {}

        for word in unique_words:
            word_counts[word] = words.count(word)

        file = open(self.output_file, "w")

        for word in word_counts:
            file.write(f"{word}: {word_counts[word]}\n")

        file.close()

        return word_counts


#### test your class
if __name__ == "__main__":
    obj = CountWordsFile("input.txt", "out.txt")
    print(obj.count_words())
```

# Task 9.
- Implement a basic chatbot that can respond to user input with predefined responses

CONVERT IT TO CLASS STRUCTURE

```python
import random

# Predefined responses
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}

# Function to get a response based on user input
def get_response(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
    
    return random.choice(responses["default"])

# Chatbot function
def chatbot():
    print("Chatbot: Hi! How can I assist you today?")
    
    while True:
        user_input = input("User: ").lower()
        response = get_response(user_input)
        print("Chatbot:", response)
        
        if user_input == "goodbye":
            break

# Run the chatbot
chatbot()
```

```python
import random
#### your class here
class ChatBot:
    """
    class ChatBot is simple chatbot with predefined responses
    Attributes:
        responses (dict): Chatbot responses
    """

    def __init__(self):
    #stores the chatbot responses
        self.responses = {
            "hello": ["Hello!", "Hi there!", "Greetings!"],
            "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
            "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
            "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
        }

    def get_response(self, user_input):
        """
        returns a chatbot response
        Returns:
            str: A response message
        """
        for key in self.responses:
            if key in user_input:
                return random.choice(self.responses[key])

        return random.choice(self.responses["default"])

    def chat(self):
        #starts the chatbot
        print("Chatbot: Hi! How can I assist you today?")

        while True:
            user_input = input("User: ").lower()
            response = self.get_response(user_input)
            print("Chatbot:", response)

            if user_input == "goodbye":
                break


#### test your class
if __name__ == "__main__":
    bot = ChatBot()
    bot.chat()
```

# Task 10: Person class and child Student class 
Create a Python class Person with attributes: name and age.
- Create a display() method that displays the name and age of an object created via the Person class.
- Create a child class Student which inherits from the Person class and which also has an extra attribute called school.
- Create a method displayStudent() that displays the name, age and section of an object created via the Student class.
- Create a student object via an instantiation on the Student class and then test the displayStudent method

```python
class Person:
    """
stores a person's info
    Attributes:
        name (str): Person's name
        age (int): Person's age
    """

    def __init__(self, name, age):
        #stores the person's name and age
        self.name = name
        self.age = age

    def display(self):
        #displays the person's info
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")


class Student(Person):
    """
    inherits from Person and stores the student's school
    Attributes:
    school (str): Student's school
    """

    def __init__(self, name, age, school):
        #stores the student's info
        super().__init__(name, age)
        self.school = school

    def displayStudent(self):
        #displays the student's info
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"School: {self.school}")

if __name__ == "__main__":
    student = Student("Engy Abdelmalak", 22, "Ain Shams University")
    student.displayStudent()
```
