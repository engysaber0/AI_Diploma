# Task1

```python
email = "Amit_ml@gmail.edu"
at_count = email.count("@")

if at_count != 1:
    print("Invalid email")
else:
    username = email.split("@")[0]
    after_at = email.split("@")[1]

    if "." not in after_at:
        print("Invalid email")
    else:
        last_dot = after_at.rfind(".")
        domain = after_at[:last_dot]

        print("Username is:", username)
        print("Domain is:", domain)

        if email.endswith(".com"):
            print("Commercial Domain")
        elif email.endswith(".edu"):
            print("Educational Domain")
        else:
            print("Other Domain")
```

# Task3

```python
message = "&&&**$gnirtS PLIO!!@1234"
clean = ""
for i in message:
    if i.isalpha() or i == " ":
        clean = clean + i
clean = clean.strip()

words = clean.split(" ")
first_word = words[0]
second_word = words[1]
reversed_word = first_word[::-1]
final_word = ""
for i in second_word:
    if i == "I":
        final_word = final_word + "E"
    elif i == "O":
        final_word = final_word + "U"
    else:
        final_word = final_word + i

print(reversed_word, final_word)
```

```python
message = "##$$$@!yalpstcejorp EPUVT****9887"
clean = ""
for i in message:
    if i.isalpha() or i == " ":
        clean = clean + i
clean = clean.strip()

words = clean.split(" ")
first_word = words[0]
second_word = words[1]

reversed_word = first_word[::-1]
final_word = ""
for i in second_word:
    if i == "E":
        final_word = final_word + "A"
    elif i == "U":
        final_word = final_word + "O"
    else:
        final_word = final_word + i

print(reversed_word, final_word)
```
