# Task 1.
- Calculate Factorial using recursion

```python
def calc_factorial(n):
    """
    function description: calculates the factorial of a number using recursion
    :param n: the number to calc factorial for
    :type n: int
    :return: the factorial of n
    :rtype: int
    """
    if n == 0:
        return 1
    if n == 1:
        return 1
    return n * calc_factorial(n - 1)

calc_factorial(3)
```

# Task 2.
- Write a function that takes a number and return True if this number is prime and False if not.

```python
def is_prime(n):
    """
    function description: checks whether n is prime
    :param n: the number to check
    :type n:int
    :return: True if the number is prime, False otherwise
    :rtype: bool
    """
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
print(is_prime(9))
```

# Task 3.
- Write a function that takes two numbers and return a list of common dividors

```python
def common_divisors(n1, n2):
    """
    function description:finds the common divisors of two numbers
    :param n1: the first number
    :type n1: int
    :param n2: the second number
    :type n2: int
    :return: a list of common divisors
    :rtype: list
    """
    divisors = []

    for i in range(1, min(n1, n2) + 1):
        if n1 % i == 0 and n2 % i == 0:
            divisors.append(i)

    return divisors


print(common_divisors(50, 25))
```

# Task 5.
- Write a function that will take a given string and reverse the order of words.

- **Ex:** “Hello world” → ”world Hello”


```python
def reverse_words(text):
    """
    function description: reverses the order of words in a string
    :param text: the input string
    :type text: str
    :return: the string with words in reverse order
    :rtype: str
    """
    words = text.split()
    words.reverse()
    return " ".join(words)
print(reverse_words("Hello world"))
```

- Create a program that generates a random password of a given length.
- Hint: use random module → import random → random.randrange

```python
import random

def generate_password(l):
    """
    function description: generates a random password of a given length
    :param l: the length of the password
    :type l: int
    :return: the generated password
    :rtype: str
    """
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789!@#$%^&*"
    password = ""

    for i in range(l):
        index = random.randrange(len(characters))
        password += characters[index]
    return password
print(generate_password(8))
```

# Task 7.
- Write a Python program to find all the unique words and count the frequency of occurrence from a given list of strings

- **Ex:** ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]  → {'Ali': 4, 'Welcome': 1, 'No': 3, 'Hi': 2}

```python
def word_frequency(words):
    """
    function description: counts the frequency of each unique word in the list
    :param words: the list of words
    :type words:list
    :return: a dictionary containing each word & its freq
    :rtype: dict
    """
    frequency = {}
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
print(word_frequency(words))
```
