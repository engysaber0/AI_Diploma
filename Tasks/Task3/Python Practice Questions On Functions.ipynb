 1. Write a function that inputs a number and prints the multiplication table of that 
number

```python
def multiply_table(n):
    """
    function description: prints the multiplication table of a number
    :param n: the number to print its multiplication table
    :type n: int
    """

    for i in range(1, 11):
        print(n, "x", i, "=", n*i)


multiply_table(9)
```

2. Write a program to print twin primes less than 1000. If two 
consecutive odd numbers are both prime then they are known as 
twin primes 


```python
def is_prime(n):
    """
    function description: checks whether n is prime
    :param n: the number to check
    :type n: int
    :return: True if the number is prime,False if not
    :rtype: bool
    """
    if n < 2:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True

print("Twin Prime:")

for i in range(3, 1000, 2):
    if is_prime(i) and is_prime(i+2):
        print(i, "and", i + 2)
```

3. Write a program to find out the prime factors of a number. 
Example: prime factors of 56 - 2, 2, 2, 7

```python
def primeFactors(n):
    """
    function description:finds the prime factors of n
    :param n:number to find prime factors for
    :type n:int
    :return: a list of prime factors
    :rtype: list
    """
    factors = []

    for i in range(2, n+1):
        while n % i == 0:
            factors.append(i)
            n = n // i

    return factors


print(primeFactors(56))
```

4. Write a function that converts a decimal number to binary 
number

```python
def decToBin(n):
    """
    function description:converts a decimal number to binary
    :param n: the decimal number
    :type n: int
    :return: the binary representation of the number
    :rtype:str
    """
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary
        n = n // 2
    return binary


print(decToBin(11))
```

5. A number is called perfect if the sum of proper divisors of that 
number is equal to the number. For example 28 is perfect 
number, since 1+2+4+7+14=28. Write a program to print all the 
perfect numbers in a given range

```python
def perfectNums(s,e):
    """
    function description:prints all perfect numbers in a given range
    :param s: the starting number
    :type s: int
    :param e: the ending number
    :type e: int
    """
    for i in range(s, e+1):
        total = 0

        for j in range(1, i):
            if i % j == 0:
                total += j

        if total == i:
            print(i)

perfectNums(0, 100)
```
